```python
"""
Project: Parameter-Efficient Federated Aggregator for Multi-Agent Optimization
Author: Kehinde Julius Adediran
Scope: PyTorch implementation of FedAvg with Parameter-Efficient (Adapter-style) updates 
       and weighted server aggregation.

Theoretical Overview:
---------------------
Federated Learning (FL) enables decentralized nodes (agents) to collaboratively train a shared model 
without transmitting private local datasets to a centralized server. Standard FL algorithms 
(e.g., McMahan et al., FedAvg) require broadcasting the full set of model parameters across edge 
networks every communication round. For large policy models, this creates severe communication bottlenecks.

This implementation incorporates Parameter-Efficient Fine-Tuning (PEFT) via bottleneck adapters. 
By freezing the deep backbone feature extractor and restricting local stochastic gradient descent (SGD) 
updates exclusively to low-rank adapter projections and the policy head, upstream bandwidth requirements 
are reduced substantially while maintaining multi-agent convergence stability.
"""

import copy
from typing import Dict, List, Tuple
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
```

```python
class AgentPolicyNetwork(nn.Module):
    """
    Local state-action policy network with parameter-efficient projection layers.

    Architectural Breakdown:
    ------------------------
    1. Feature Extractor (Backbone): Multi-layer perceptron (MLP) responsible for mapping high-dimensional
       state inputs into latent space representations. Designed to be frozen during local PEFT passes.
    2. Bottleneck Adapter: Low-rank projection module inserted before the output layer. Down-projects 
       latent dimensions to a bottleneck size (hidden_dim // 4) before projecting back up, applying 
       a non-linear activation in between.
    3. Residual Connection: Adds original latent features to adapter output, ensuring initial identity-like 
       mapping during early training phases.
    4. Policy Head: Final mapping layer converting adapted latent features into action logits.
    """

    def __init__(self, input_dim: int = 64, hidden_dim: int = 128, action_dim: int = 10):
        """
        Initializes the AgentPolicyNetwork architecture.

        Args:
            input_dim (int): Dimensionality of input state vector. Default: 64.
            hidden_dim (int): Hidden dimension size for internal representations. Default: 128.
            action_dim (int): Dimensionality of discrete action output logits. Default: 10.
        """
        super().__init__()
        
        # Core feature extractor backbone (Representation Learning Phase)
        self.feature_extractor = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
        )
        
        # Parameter-efficient bottleneck adapter module (PEFT Phase)
        # Reduces parameter count during local updates by constricting latent bottleneck size.
        self.adapter = nn.Sequential(
            nn.Linear(hidden_dim, hidden_dim // 4, bias=False),  # Down-projection
            nn.ReLU(),                                           # Non-linear bottleneck activation
            nn.Linear(hidden_dim // 4, hidden_dim, bias=False),  # Up-projection
        )
        
        # Final state-action policy mapping layer
        self.policy_head = nn.Linear(hidden_dim, action_dim)

    def freeze_backbone(self) -> None:
        """
        Freezes feature extractor parameters for parameter-efficient fine-tuning (PEFT).

        Implementation Detail:
        ----------------------
        Sets `requires_grad = False` across all tensors within `self.feature_extractor`.
        This prevents autograd engine from computing or storing gradients for the backbone 
        during local backpropagation passes, saving compute cycles and memory overhead.
        """
        for param in self.feature_extractor.parameters():
            param.requires_grad = False

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        Executes forward pass with residual adapter modification.

        Mathematical Formulation:
        -------------------------
        h = FeatureExtractor(x)
        h_adapted = h + Adapter(h)   [Residual Connection]
        logits = PolicyHead(h_adapted)

        Args:
            x (torch.Tensor): Input batch tensor of shape (batch_size, input_dim).

        Returns:
            torch.Tensor: Unnormalized policy action logits of shape (batch_size, action_dim).
        """
        features = self.feature_extractor(x)
        # Residual connection ensures adapter acts as a perturbation field over frozen representations
        adapted_features = features + self.adapter(features)
        return self.policy_head(adapted_features)


class FederatedAgentNode:
    """
    Decentralized edge node fine-tuning local policy via parameter-efficient updates.

    Responsibilities:
    -----------------
    - Holds local private datasets that are never exposed to external clients or central servers.
    - Manages local model replica instances synchronized with global state snapshots.
    - Executes isolated optimization rounds (e.g., local mini-batch SGD).
    - Extracts and packages only trainable parameter weights for efficient communication overhead.
    """

    def __init__(
        self,
        node_id: int,
        model: nn.Module,
        data_loader: DataLoader,
        lr: float = 0.01,
        peft_only: bool = True,
    ):
        """
        Initializes an isolated local client node within the federated network.

        Args:
            node_id (int): Unique identifier for logging and node tracking.
            model (nn.Module): Reference global model template to deepcopy locally.
            data_loader (DataLoader): PyTorch DataLoader holding private local node data.
            lr (float): Local optimizer learning rate. Default: 0.01.
            peft_only (bool): If True, freezes backbone and restricts optimization to PEFT layers. Default: True.
        """
        self.node_id = node_id
        # Instantiate an isolated local copy of global policy model to prevent memory reference leaks
        self.local_model = copy.deepcopy(model)
        self.data_loader = data_loader
        self.num_samples = len(data_loader.dataset)

        # Apply PEFT parameter constraints if flagged
        if peft_only:
            self.local_model.freeze_backbone()

        # Isolate trainable parameters (adapters + policy head) to exclude frozen backbone from SGD state
        trainable_params = [p for p in self.local_model.parameters() if p.requires_grad]
        self.optimizer = optim.SGD(trainable_params, lr=lr)
        self.criterion = nn.CrossEntropyLoss()

    def train_local_epoch(self, local_epochs: int = 3) -> Tuple[Dict[str, torch.Tensor], int]:
        """
        Executes local SGD optimization epochs over private node data.

        Procedure:
        ----------
        1. Set local model to train mode.
        2. Iterate through local mini-batches over specified `local_epochs`.
        3. Zero optimizer gradients, calculate cross-entropy loss, and step parameters.
        4. Extract state dictionary containing *only* parameters with `requires_grad == True`.

        Args:
            local_epochs (int): Number of local training sweeps over local data loader. Default: 3.

        Returns:
            Tuple[Dict[str, torch.Tensor], int]: 
                - Trainable state dictionary payload containing updated local weights.
                - Total sample size ($n_k$) used for server-side weighted aggregation calculations.
        """
        self.local_model.train()
        
        # Local mini-batch optimization loops
        for _ in range(local_epochs):
            for data, targets in self.data_loader:
                self.optimizer.zero_grad()
                output = self.local_model(data)
                loss = self.criterion(output, targets)
                loss.backward()
                self.optimizer.step()

        # Extract only parameters that require gradients (Adapter + Policy Head)
        # Detach and clone tensors off execution graph to prevent memory leakage and wire serialization issues.
        trainable_state_dict = {
            k: v.cpu().detach().clone()
            for k, v in self.local_model.state_dict().items()
            if self.local_model.get_parameter(k).requires_grad
        }
        return trainable_state_dict, self.num_samples


class FederatedServer:
    """
    Central Orchestrator administering sample-weighted FedAvg over client updates.

    Mathematical Definition (FedAvg Aggregation):
    ----------------------------------------------
    Given $K$ participating client nodes where node $k$ processes $n_k$ local samples, 
    let the total dataset size across participating nodes be:
        $N = \sum_{k=1}^{K} n_k$

    The global parameter aggregation for parameter vector $\theta$ at round $t+1$ is:
        $\theta^{t+1} = \sum_{k=1}^{K} \frac{n_k}{N} \theta_k^{t+1}$

    In this PEFT regime, $\theta$ represents only the trainable subset (Adapters + Head).
    """

    def __init__(self, global_model: nn.Module):
        """
        Initializes central server managing global reference model.

        Args:
            global_model (nn.Module): Master parameter reference model.
        """
        self.global_model = global_model

    @torch.no_grad()
    def aggregate_updates(
        self, client_updates: List[Tuple[Dict[str, torch.Tensor], int]]
    ) -> None:
        """
        Applies parameter-weighted federated averaging (FedAvg) directly in memory.

        Memory Optimization:
        --------------------
        Performs inplace tensor summation into pre-allocated accumulator buffers without 
        stacking full node state dictionaries in GPU/CPU RAM simultaneously.

        Args:
            client_updates (List[Tuple[Dict[str, torch.Tensor], int]]): List of tuples containing:
                - Client state dictionary payload (trainable parameters only).
                - Client local sample count ($n_k$).

        Raises:
            ValueError: If aggregate sample count across all clients equals zero.
        """
        # Calculate global sample count denominator (N = sum(n_k))
        total_samples = sum(num_samples for _, num_samples in client_updates)
        if total_samples == 0:
            raise ValueError("Total sample count across clients must be greater than zero.")

        global_dict = self.global_model.state_dict()

        # Extract target keys present in client updates (supports full or PEFT partial dict updates)
        update_keys = client_updates[0][0].keys()

        # Allocate zero-initialized memory accumulation buffers matching target global tensor dimensions
        aggregated_weights: Dict[str, torch.Tensor] = {
            key: torch.zeros_like(global_dict[key], dtype=torch.float32) for key in update_keys
        }

        # Accumulate weighted client parameter contributions: sum((n_k / N) * theta_k)
        for client_dict, num_samples in client_updates:
            weight = num_samples / total_samples  # Calculate relative client weight factor
            for key in update_keys:
                aggregated_weights[key] += client_dict[key].to(dtype=torch.float32) * weight

        # Copy aggregated weighted parameters back into global model state dictionary
        for key, tensor in aggregated_weights.items():
            global_dict[key].copy_(tensor)

        # Reload updated global state dict back into global PyTorch module
        self.global_model.load_state_dict(global_dict)


if __name__ == "__main__":
    # =========================================================================
    # 1. INITIALIZATION & SETUP
    # =========================================================================
    # Instantiate global master policy model on central server
    global_policy = AgentPolicyNetwork()
    server = FederatedServer(global_policy)

    # Simulate unbalanced/non-IID dataset size distribution across 3 edge clients
    # Client 0: 64 samples | Client 1: 128 samples | Client 2: 32 samples
    clients: List[FederatedAgentNode] = []
    sample_sizes = [64, 128, 32]

    # Instantiate simulated client dataset loaders
    for i, n_samples in enumerate(sample_sizes):
        # Generate synthetic state input tensors (n_samples x 64 feature dimensions)
        synthetic_x = torch.randn(n_samples, 64)
        # Generate synthetic target label tensors (discrete action space 0-9)
        synthetic_y = torch.randint(0, 10, (n_samples,))
        
        dataset = TensorDataset(synthetic_x, synthetic_y)
        loader = DataLoader(dataset, batch_size=16, shuffle=True)

        # Initialize local edge node bound to synthetic dataset loader
        client = FederatedAgentNode(
            node_id=i, model=global_policy, data_loader=loader, peft_only=True
        )
        clients.append(client)

    print("[Federated Agent Training Initialized]")

    # =========================================================================
    # 2. FEDERATED COMMUNICATION LOOPS
    # =========================================================================
    num_communication_rounds = 3

    for round_num in range(1, num_communication_rounds + 1):
        # Buffer to store node update payloads submitted during current communication round
        client_updates: List[Tuple[Dict[str, torch.Tensor], int]] = []

        # Broadcast current global master model parameters to all edge clients
        global_state = server.global_model.state_dict()
        
        for client in clients:
            # Synchronize local model parameters with broadcast global state dict.
            # strict=False allows partial state dict synchronization (PEFT parameters).
            client.local_model.load_state_dict(global_state, strict=False)

            # Perform isolated local training iterations over client's private data
            updated_weights, num_samples = client.train_local_epoch(local_epochs=2)
            
            # Stage client weight payload and sample volume metadata for central aggregation
            client_updates.append((updated_weights, num_samples))

        # Perform parameter-weighted global aggregation (FedAvg) at central server
        server.aggregate_updates(client_updates)
        
        print(
            f"Round {round_num}: Aggregation complete across {len(clients)} nodes "
            f"(Total processed samples: {sum(s for _, s in client_updates)})."
        )
```
